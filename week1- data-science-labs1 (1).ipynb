```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

data = {
    "age": [20, 35, 40, 25, 50, 30, 45, 28],
    "monthly_bill": [20, 80, 90, 30, 100, 40, 95, 25],
    "complaints": [0, 3, 4, 0, 5, 1, 4, 0],
    "churn": [0, 1, 1, 0, 1, 0, 1, 0]
}

df = pd.DataFrame(data)


X = df[["age", "monthly_bill", "complaints"]]
y = df["churn"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)


model = DecisionTreeClassifier()
model.fit(X_train, y_train)


prediction = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, prediction))
```

## Data Science Workflow

A **data science workflow** is a systematic process of using data to understand problems and develop effective business solutions.

1. **Define the Problem** – Determine the business issue and establish the main objective.
2. **Gather Data** – Obtain useful information from sources such as databases, surveys, websites, and applications.
3. **Prepare the Data** – Clean the dataset by correcting errors and dealing with missing or repeated values.
4. **Explore the Data** – Examine the data to discover important trends, patterns, and relationships.
5. **Select Features** – Choose the variables that are most relevant to solving the problem.
6. **Develop a Model** – Use statistical or machine learning techniques to create a predictive model.
7. **Evaluate the Model** – Measure how well the model performs and determine its accuracy.
8. **Deploy the Model** – Implement the solution in an actual business environment.
9. **Monitor the Results** – Track the model's performance and update it when necessary.

### Real Business Problem: Customer Churn Prediction

A **telecommunication company** may want to predict which customers are likely to stop using its services.

The company can analyze data such as:

* Customer age
* Monthly payment
* Internet usage
* Number of complaints
* Type of contract
* Previous customer behavior

A machine learning model can classify customers as:

**0 → Customer is likely to stay**
**1 → Customer is likely to leave**

Using these predictions, the company can provide **special offers, discounts, improved plans, or better customer support** to customers who are at risk of leaving.

**Business Benefit:** This approach can help the company **retain more customers, reduce financial losses, increase revenue, and improve customer satisfaction**.
